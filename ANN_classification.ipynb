```python
X = df.drop("Class" , axis = 1)
Y = df["Class"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler, LabelEncoder

le = LabelEncoder()
Y = le.fit_transform(Y)
```

```python
Y
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

ANN

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
```

```python
#yaha for Y train/test we use torch.long datatype because as for multiclass - classifiation in 
#neural network, the loss function that we use is Cross Entropy loss function which expedcts labels/target values in long dtype
#pytorch se hum jab NN build krte hai , multiclass cross entropy use krte hai
```

```python
X_train_tensor = torch.tensor(X_train_scaled, dtype = torch.float32)
Y_train_tensor = torch.tensor(Y_train, dtype = torch.long)

X_test_tensor = torch.tensor(X_test_scaled, dtype = torch.float32)
Y_test_tensor = torch.tensor(Y_test, dtype = torch.long)
```

```python
train_dataset = TensorDataset(X_train_tensor,  Y_train_tensor)
test_dataset = TensorDataset(X_test_tensor,  Y_test_tensor)
```

```python
#train loader k liye shuffle krna is necessary because when we split, kahi aisa na ho jaye some values mile hi nhi for training set.
#upar check Y. 0,1,2.. all in sequence. so we shuffle it to get better training set
train_loader = DataLoader(train_dataset, batch_size=32, shuffle = True)
test_loader = DataLoader(test_dataset, batch_size = 32)
```

```python
#Build Model

class ANN(nn.Module):
    def __init__(self):
        super(ANN, self).__init__()

        self.model = nn.Sequential(
            nn.Linear(X.shape[1], 64),
            nn.ReLU(),
            nn.Linear(64, 64),
            nn.ReLU(),
            nn.Linear(64, 7 )
        )
        #yaha pe output k liye we have 7 layers becoz, we have total 7 classes as our labels (the "CLASS" column)

    def forward(self, x):
        return self.model(x)
```

```python
model = ANN()

#Loss and optimizer
criteria = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters())
```

```python
#database chota hai isslie we aren't using validation

epochs = 100

#jab model ko training mode mai dalna ho, we do "model.train()" aur jab validate & evaluate krna ho, we do "model.eval() , for both modes, we define it again "
for epoch in range(epochs):
    model.train()

    running_loss = 0.0

    for xb,  yb in train_loader:
        optimizer.zero_grad()           #optimizer prev gradients ko bhi accumulate krta hai, so in order to avoid that, we use this line.
        
        outputs = model(xb)             #forward propagation
        loss = criteria(outputs, yb)   #loss & gradient computation
        loss.backward()                 #backward propagation
        optimizer.step()                #weight/learnable params update krne k lie 
        running_loss += loss

    train_loss = running_loss / len(train_loader)     #this is the average loss per batch in 1 epoch.  (sare loss ko we divide by len of sare losses)

    print(f"epoch = {epoch+1}/{epochs},  loss = {train_loss}")
```

```python
#evaluate
model.eval()

total = 0
correct = 0

with torch.no_grad():
    for xb, yb in test_loader:
        outputs = model(xb)
        max_val, predicted = torch.max(outputs, 1 )     #this returns max value and its index position which we store simultaneously

        correct += (predicted == yb).sum().item()  #jitne bhi predicted k values are equal to actual labels, unn sab ka sum karo, convert it to python and store it for all batches
        total += yb.size(0)                        #actual samples in each batch

print("Correct vals :", correct)
print("Total vals :", total)

print("Accuracy :", correct/total * 100)
```
